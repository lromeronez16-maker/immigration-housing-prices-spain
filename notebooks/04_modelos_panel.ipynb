# 04 · Datos de panel con efectos fijos

$$\Delta \ln P_{pt}\times 100 = \beta\, m_{pt} + X_{pt}'\gamma + \mu_p + \tau_t + \varepsilon_{pt}$$

- $\mu_p$: efectos fijos provinciales (tendencias de precio específicas de cada provincia).
- $\tau_t$: efectos fijos de año (ciclo nacional, tipos de interés, crédito, política monetaria).
- Errores agrupados por provincia.

Incluye flujo retardado, retardos distribuidos (efecto acumulado) y contrastes de Wald.
